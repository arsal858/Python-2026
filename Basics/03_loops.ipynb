**What are Loops**  
Loops are the powerful way of repeating a piece of code again and again.

```python
i=3
while i!=0:
    print('hy')
    i=i-1
```

```python
i=5
while i!=0:
    print('Arsal')
    i=i-1
```

```python
while True:
    n=int(input('Enter the Number'))
    if n<0:
        continue
    else:
        break
for _ in range(n):
    print('Arsaln Shahzad Gondal')
```

```python
i=0
while i<=3:
    print('buy')
    i+=1
```
